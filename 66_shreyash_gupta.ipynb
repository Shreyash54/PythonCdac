```python
#Question 1
def longest_increasing_subsequence(arr):
   if not arr:
       return 0
   n = len(arr)
   lis = [1] * n 
   for i in range(1, n):
       for j in range(i):
           if arr[i] > arr[j] and lis[i] < lis[j] + 1:
               lis[i] = lis[j]+1
   return max(lis)

def get_lis_elements():
    newarr=[]
    n=len(arr)
    lis=[1]*n
    for i in range( 1,n):
       for j in range(i):
           if arr[i]>arr[j] and lis[i]<lis[j]+1 :
               lis[i]=lis[j]+1
               lis[i]=arr[i]
               newarr.append(lis[i])
           return newarr

arr = input("enter the list :").split(",")
print("Length of listis", longest_increasing_subsequence(arr))
print("Longest subsequnce is " ,get_lis_elements())
            
```

```python
#Question 2
def fibonacci_modified(t1,t2,n):
    if n==0:
        return t1
    elif n==1:
        return t2
    else :
        return fibonacci_modified(t1,t2,n-1)+fibonacci_modified(t1,t2,n-2)
        
def nthterm(t1,t2,n):
    seq=[t1,t2]
    for i in range(2,n):
        term= seq[i-2] +seq[i-1]**2
        seq.append(term)
    return seq[n-1]
        
def sum_range(t1,t2,start,end):
    sum=0
    for i in range(start,end):
        sum +=fibonacci_modified(t1,t2,i)
    return sum
def count_divisible(t1,t2,n,k):
    count=0
    for i in range(1,n):
        if fibonacci_modified(t1,t2,i)%k==0:
            count+=1
    return count
t1,t2=0,1
start,end =1,5
n=5
k=2
print("The nth fibonacci terms is :",nthterm(t1,t2,n))
print(f"the sum from {start} - {end} is :",sum_range(t1,t2,start,end))
print("the count ofnumbers divisble by k is ",count_divisible(t1,t2,n,k))
```
